```python
for animal in animal_ID[-2:-1]:
    exp_day = os.listdir(os.path.join(data_path, animal))
    for day in exp_day:
        exp_session = os.listdir(os.path.join(data_path, animal, day))
        for session in exp_session:
            session_file = os.listdir(os.path.join(data_path, animal, day, session))
            for file in session_file:
                session_path = os.path.join(data_path, animal, day, session)
                extension = str.split(file, ".")[1]
                if "cam_23175601" in file:
                    os.rename(os.path.join(session_path, file), os.path.join(session_path, animal + "_pov_face." + extension))
                if "cam_23199174" in file:
                    os.rename(os.path.join(session_path, file), os.path.join(session_path, animal + "_pov_tunnel." + extension))
```

```python
import os
import shutil

def move_and_rename_h5(base_dir):
    for day_name in os.listdir(base_dir):
        day_path = os.path.join(base_dir, day_name)

        for condition_name in os.listdir(day_path):
            condition_path = os.path.join(day_path, condition_name)

            for file in os.listdir(condition_path):
                if file.endswith(".h5"):
                    src = os.path.join(condition_path, file)

                    new_filename = f"social_{condition_name}.h5"
                    dst = os.path.join(day_path, new_filename)

                    shutil.move(src, dst)
                    print(f"Moved: {src} -> {dst}")
```

```python
base_dir = 'C:/Users/esay/data/social_interaction/SLEAP/SLEAP_raw/shank3_new/social-8564-2'

move_and_rename_h5(base_dir)
```
